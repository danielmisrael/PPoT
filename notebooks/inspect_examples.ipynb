```python
temperature = 0.7
num_examples = 10
num_samples = 1000
isUSPP = False
greedy = True
pp_temperature = 1.0

with open(f"/space/poorvagarg/genPPS/out/t{pp_temperature:.1f}/stats_t{temperature:.1f}_n{num_examples}_s{num_samples}"
            + isUSPP*"_isUSPP" + greedy*"_greedy" + "_examples" + ".pkl", "rb") as f:
    triples = pickle.load(f)
```

```python
i = 0
P = triples[i]
os.makedirs(f"examples/{i}", exist_ok=True)
print("--------------------------------------------------")
print("Sampled Program")
with open(f"examples/{i}/sample.py", "w") as f:
    f.write(P[0])

print("--------------------------------------------------")
print("Raw Program")
with open(f"examples/{i}/raw.py", "w") as f:
    f.write(P[1][0])

print("--------------------------------------------------")
print("Ground Truth")
with open(f"examples/{i}/truth.py", "w") as f:
    f.write(P[2])
```
