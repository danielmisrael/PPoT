```python
import argparse, pickle, os
from ppot.utils import safe_execute_plot
from scripts.text_match_score import evaluate_single_example
```

```python
temperature = 0.7
num_examples = 10
num_samples = 10
isUSPP = False
greedy = True
pp_temperature = 1.0

with open(f"/space/poorvagarg/genPPS/out/t{pp_temperature:.1f}/stats_t{temperature:.1f}_n{num_examples}_s{num_samples}"
            + isUSPP*"_isUSPP" + greedy*"_greedy" + "_examples" + ".pkl", "rb") as f:
    triples = pickle.load(f)
```

```python
i = 0
P = triples[i]
os.makedirs(f"examples/{i}", exist_ok=True)
print("--------------------------------------------------")
print("Sampled Program")
with open(f"examples/{i}/sample.py", "w") as f:
    f.write(P[0])

print("--------------------------------------------------")
print("Raw Program")
with open(f"examples/{i}/raw.py", "w") as f:
    f.write(P[1][0])

print("--------------------------------------------------")
print("Ground Truth")
with open(f"examples/{i}/truth.py", "w") as f:
    f.write(P[2])
```

```python
sample = open(f"examples/{i}/sample.py").read()
raw = open(f"examples/{i}/raw.py").read()
truth = open(f"examples/{i}/truth.py").read()
success_sample, result_sample = safe_execute_plot(sample, timeout_seconds=10, separate_process=True)
success_raw, result_raw = safe_execute_plot(raw, timeout_seconds=10, separate_process=True)
success_truth, result_truth = safe_execute_plot(truth, timeout_seconds=10, separate_process=True)
```
